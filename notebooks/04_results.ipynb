# 4. Results: the tables and figures of the paper

`RESULTS` is the `results/` folder of the public data (computed with `analysis.py` and the released
checkpoints), so this notebook needs no GPU. To recompute everything yourself, run the last cell
(several hours) and point `RESULTS` at its output folder.

```python
import sys
from pathlib import Path

sys.path.insert(0, "..")          # the repository root, one level up from notebooks/
DATA_ROOT = Path("../data")       # the public data: manifests/, frames/, stacks/, pairs/, checkpoints/, results/
import csv
import analysis, figures
RESULTS = DATA_ROOT / "results"
SAVE = Path("my_results")         # where tables and figures are saved below
```

## Tables

```python
def show_table(name, save_as=()):
    """Print tables/<name>.csv and optionally save it again as .csv and/or .txt under SAVE."""
    rows = list(csv.reader(open(RESULTS / "tables" / f"{name}.csv")))
    title = open(RESULTS / "tables" / f"{name}.txt").readline().strip()
    if save_as:
        analysis.write_table(SAVE, name, rows[0], rows[1:], title)   # writes SAVE/tables/<name>.csv and .txt
    else:
        print((RESULTS / "tables" / f"{name}.txt").read_text())

show_table("table1")
```

```python
show_table("table2")
```

```python
show_table("table1", save_as=("csv", "txt"))   # saved to my_results/tables/
```
