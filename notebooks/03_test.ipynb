# 3. Testing a checkpoint

`test.py` computes the model rows of Tables 1 and 2. The functions below are the ones the command
line uses (`python test.py --ckpt <model.pt> --data-root <data> --table 1a 1b 1c 2a 2b`).
Table 1a takes a few minutes on a GPU; Tables 1b/1c (stacking of the frames) and 2a (2,468 pairs)
take longer.

```python
import sys
from pathlib import Path

sys.path.insert(0, "..")          # the repository root, one level up from notebooks/
DATA_ROOT = Path("../data")       # the public data: manifests/, frames/, stacks/, pairs/, checkpoints/, results/
import torch
import test
dev = torch.device("cuda" if torch.cuda.is_available() else "cpu")
CKPT = DATA_ROOT / "checkpoints" / "n2n_mixsigma_s1" / "model.pt"
methods = {"input": lambda b: b, "n2n_mixsigma": test.load_denoiser(CKPT, dev)}
```

## Table 1(a): the 21-frame stack reference

```python
rows_1a = test.run(["1a"], methods, DATA_ROOT, splits=["test"], out_dir="results_test")["1a"]
```

## Tables 2(a)-(c): HMI-MDI and HMI-PHI

```python
rows_2 = test.run(["2b"], methods, DATA_ROOT, out_dir="results_test")   # 14 HMI-PHI epochs, quick
```

```python
rows_2a = test.run(["2a"], methods, DATA_ROOT, out_dir="results_test")  # 2,468 HMI-MDI pairs
```

Each table is also written as `results_test/table<t>_rows.csv` (one row per group or pair).
